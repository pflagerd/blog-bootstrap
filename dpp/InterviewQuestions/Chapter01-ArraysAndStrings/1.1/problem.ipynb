cannot use additional data structures?

```python
def isUnique(s):
    if s is None or len(s) == 0:
        return False
    
    left = None
    right = None
    for c in sorted(s):
        print(type(c))
        if left is None:
            left = c
            continue

        if right is None:
            right = c
        else:
            left = right
            right = c        

        if left != right:
            return False

    return True
```

```python
import unittest

class TestCase(unittest.TestCase):
    def test_A(self):
        isUnique("The quick brown fox jumps over the lazy dog")

if __name__ == "__main__":
    unittest.main(argv=[""], verbosity=2, exit=False)
```
