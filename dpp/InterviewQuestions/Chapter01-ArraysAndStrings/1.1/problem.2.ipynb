**Is Unique:** Implement an algorithm to determine if a string has all unique characters. What if you
cannot use additional data structures?
    
**Hints**:  
&emsp;#44   Try a hash table.  
&emsp;#117  Could a bit vector be useful?  
&emsp;#132  Can you solve it in O(N log N) time? What might a solution like that look like?  

```python
def isUnique(s):
    hashTable = [False] * 256
    for c in s:
        if hashTable[ord(c)]:
            return False

        hashTable[ord(c)] = True

    return True
```

```python
import unittest

class TestCase(unittest.TestCase):
    def testA(self):
        self.assertFalse(isUnique("The quick brown fox jumps over the lazy dog."))

if __name__ == "__main__":
    unittest.main(argv = [""], verbosity = 2, exit = False)
```
