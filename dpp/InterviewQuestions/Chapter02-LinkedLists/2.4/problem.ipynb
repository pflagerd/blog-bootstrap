**2.4 Partition:** Write code to partition a linked list around a value x, such that all nodes less than x come
before all nodes greater than or equal to x. If x is contained within the list, the values of x only need
to be after the elements less than x (see below). The partition element x can appear anywhere in the
"right partition"; it does not need to appear between the left and right partitions.

EXAMPLE

Input:  3 -> 5 -> 8 -> 5 -> 10 -> 2 -> 1 [partition=5]

Output: 3 -> 1 -> 2 -> 10 -> 5 -> 5 -> 8

*Hints: #3, #24*


```python
class SinglyLinkedListNode:
    # <code>value</code> is passed the integer payload to be contained by the new <code>SinglyLinkedListNode</code>.
    def __init__(self, value: int, next: SinglyLinkedListNode | None = None) -> None:
        self.next = next
        self.value = value

    def __eq__(self, other):
        if not isinstance(other, SinglyLinkedListNode):
            return False

        current_self = self.next
        current_other = other.next

        # Traverse both lists simultaneously
        while current_self and current_other:
            if current_self.value != current_other.value:
                return False
            current_self = current_self.next
            current_other = current_other.next

        # If both are None, they are the same length and identical
        return current_self is None and current_other is None


    def __repr__(self) -> str:
        s = f"SinglyLinkedListNode({self.value}"
        if self.next is not None:
            s += ", " + repr(self.next)
        s += ")"
        return s


    def dumps(self) -> str:
        s = '{"next": '
        if self.next is None:
            s += "null"
        else:
            s += str(self.next.dumps())
        s += ", \"value\": " + str(self.value)
        s += "}"
        return s
```

```python
#
# Contradictions?
# Ambiguities?
#    Node vs value
#    Node vs list (sublist)
#
# Daniel and David notices he constructs "linked lists with the head on the left and the tail on the right".
#
# Partition a linked list around a value x, such that all nodes whose value is less than x come before all nodes whose value is greater than or equal to x. 
# If x is contained within the list, the values of x only need to be after the elements less than x (see below). 
# A partition element x can appear anywhere in the "right partition"; it does not need to appear between the left and right partitions.
#
# EXAMPLE
#   Input: 3 -> 5 -> 8 -> 5 -> 10 -> 2 -> 1 [partition=5]
#   Output: 3 -> 1 -> 2 -> 10 -> 5 -> 5 -> 8
#
# Notice that the problem statement has defined some terms (objects) for us: "x", "left partition" and "right partition".
#   NOTE: Double quotes "Q" above mean "the object from the problem statement which she refers to as Q". i.e. When Gayle says "Q" she means "the object from the problem statement which I call Q".
# Notice that we are assuming some definitions:
#   "linked list" is more precisely called as a "singly linked list" which we know from information in the chapter of the Cracking the Coding Interview we're in: Chapter 2.
#   A "singly linked list" is an ordered list of "node"s.
#   A "singly linked list" is identified by a pointer, typically called "head", pointing to its first "node".
#   A "node" contains (has a) a pointer, typically called "next", to the next "node" or it contains the value null (None in Python) if it is the last "node" of the list.
#   The last "node" of the list is also known as the list's "tail".
#   A "node" also contains (has a) a value, typically called "value"
#
# Daniel "undefines" the term he invented "partition element" as an alias for "x", because Ockham's razor helps conserve mental resources here.
# Daniel made the mistake in an attempt to reduce verbage. He'd rather not have to say "value greater than or equal to x" or "value less than x", but say a one or two-word alias for either/both of these.
# Daniel made the mistake before of calling a sub-list's head as a node.  A sub-list's head IS a node, but not a SINGLE node (which is an English ambiguity).
# David and Daniel both realize that there is no unique single solution to every problem instance covered by the problem statement.
#
# Because of context, we may assume that a "node"'s "value" is an integer, although any type which has concept of "greater than or equal to" could be used.
# Hereinafter, we're dropping the double-quotes.
# My general notion of a solution is to find the first value greater than or equal to x, 
#   then for each (single) node after it
#     if that node is less than the first value greater than or equal to x, move that node such that it becomes the new head - with its <code>.next</code> pointing to the old head node.
#     This effectively makes that node the new head node.
#     if that node is greater than x it can stay where it is, because it's already "in the right partition".
#
# Desired end state: All nodes with value less than x
#
def partition(head: SinglyLinkedListNode, x: int) -> SinglyLinkedListNode:
    predecessor = None # there is no predecessor to head
    p = head # p stands for "pointer" meaning "pointer to head of sublist <condition>" where condition is to be clarified. come back to this
    while True: # looking for the first value greater than or equal to x
        if p.value < x:
            predecessor = p # p becomes its predecessor after we're done with it.
            p = p.next            
            if p is None:
                return head # come back to this. is this correct? What if no node has a value equal to x?
        else:
            break

    # p points to a sub-list's head who's value is greater than or equal to x
    # predecessor points to p
    
    # if it is equal to x, can I leave it where it is and move on?  I GUESS YES, because Gayle's problem statement says 'A partition element x can appear anywhere in the "right partition"; it does not need to appear between the left and right partitions.'
    # if it is equal to x, there may be more values like it after it in the list.  What to do with them?
    # if it is greater than x, can I leave it where it is and move on? I GUESS YES!
    c = p # c stands for "current" meaning "pointer to the node currently under investigation"
    while True:
        if c is None:
            return head

        print("predecessoor == " + str(predecessor))
        print("c == " + str(c))
        if c.value < x: # does c belong in the left partition?
            # delete node
            # deleting the a node means replacing it's predecessor's .next with node's next
            predecessor.next = c.next
            
            # disconnect node from list
            c.next = None
            
            # insert node before current head (could be inserted before the first node with node.value == x).
            # inserting a node before current head seems easier
            c.next = head
            head = c

            c = predecessor.next
            continue
            
        predecessor = c
        c = c.next            
        
    return head
```
